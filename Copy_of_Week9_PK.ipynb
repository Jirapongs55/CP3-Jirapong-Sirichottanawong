```python
import pandas as pd
import plotly.express as px
import seaborn as sns

df = pd.read_csv('https://github.com/chris1610/pbpython/blob/master/data/cereal_data.csv?raw=True')
df.head()
```

#### Reference: https://plotly.com/python-api-reference/generated/plotly.express.scatter

```python
fig = px.scatter(df,
                x='sugars',
                y='rating',
                #hover_name='name',
                color = 'shelf',
                size = 'cups',
                hover_data=['calories','carbo'],
                hover_name= df['name'].astype(str)+" : "+df['mfr'].astype(str),
                title='Cereal_ratings VS Sugars')
fig.show()
```

```python
fig = px.treemap(df, path=['mfr', 'shelf','calories','name'], values='cereal', title='Cereals by shelf location')   #title is information
fig.show()
```

```python
df = pd.read_csv("/content/cars.csv",sep = ";")
```

```python
df = df.loc[1:]
```

```python
df.describe(include="all")
```

```python
results = duckdb.sql("select * from df").df()
df.loc[:, "MPG":"Model"] = df.loc[:, "MPG":"Model"].apply(pd.to_numeric, errors="coerce")
display(results)
```

```python
results2 = duckdb.sql("select Origin, avg(MPG) from df group by Origin order by avg(MPG) desc")
display(results2)
```

```python
results3 = duckdb.sql("select Cylinders, avg(Displacement) from df group by Cylinders order by avg(Displacement) desc")
display(results3)
```

```python
display(df.loc[:, "Model"].unique())
```

```python
results4 = duckdb.sql("select Origin, Model, avg(Weight), avg(Cylinders) from df group by Origin, Model order by avg(Weight) asc")
display(type(results4))
display(results4)
```

```python
# Graph by Plotly
df4 = results4.fetchdf()
fig = px.scatter(df4,
                 x = df4.iloc[:, 1], y=df4.iloc[:,2],
                 color="Origin", size="avg(Cylinders)",
                 text="Origin", symbol = "Origin")
fig.update_layout(
    xaxis_title="Model_year",
    yaxis_title="Weight",
    title="Car weight vs. model year by origin",
)
fig.show()
```
